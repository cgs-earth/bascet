```python
from censusdis.states import STATE_CO
import os
from dotenv import load_dotenv
import censusdis.data as ced

from basket.access_cps import make_recode_info, combine_recode_info, master_cps

%load_ext autoreload
%autoreload 2
```

```python
load_dotenv()
c = os.getenv("CENSUS_API_KEY")
```

```python
YEAR = 2025
DATASET = "cps/asec/mar"

ced.variables.all_variables(
    DATASET, YEAR, None
    ).query(
        'VARIABLE.str.contains("H_NUMPER|HTOTVAL", regex=True)'
        )
```

```python
hh_inc_chunks = make_recode_info(
    "HTOTVAL",
    "HH_INC",
    [
        -999999,  # total min
        10000,  # category min
        20000,  # category min
        30000,  # category min
        40000,  # category min
        50000,  # category min
        60000,  # category min
        75000,  # category min
        100000,  # category min
        125000,  # category min
        150000,  # category min
        200000,  # category min
        99999999,  # total max
    ],
)
```

```python
hh_size_chunks = make_recode_info(
    "H_NUMPER",
    "HH_SIZE",
    [
        1,  # total min
        2,  # category min
        3,  # category min
        4,  # category min
        5,  # category min
        10, # category min
        20,  # category min
        39,  # total max
    ],
)
```

```python
recoded_cols = combine_recode_info([hh_inc_chunks, hh_size_chunks])
```

```python
hh_inc_size = master_cps(
    var_names=["HH_SIZE", "HH_INC"],
    recode_info=recoded_cols,
    state=STATE_CO,
    counties=False
)
```

```python
hh_inc_size.to_csv(r"hh_inc_size_2025.csv")
```
